## Loading dataset

```python
import os

file_paths = [
    "/kaggle/input/datasets/jashanpreetsingh645/agricultural-production-statistics/agricultural-production-statistics-jun02-final-tables-dairy-herd-size-region/1-dairy-region.xls",
    "/kaggle/input/datasets/jashanpreetsingh645/agricultural-production-statistics/agricultural-production-statistics-jun02-final-tables-pigs-number-region/5-pigs-region.xls",
    "/kaggle/input/datasets/jashanpreetsingh645/agricultural-production-statistics/agricultural-production-statistics-jun17-final-tables/Copy of agricultural-production-statistics-jun17-final-tables-v2.xlsx",
    "/kaggle/input/datasets/jashanpreetsingh645/agricultural-production-statistics/agricultural-production-statistics-jun02-final-tables-deer-number-region/5-deer-region.xls",
    "/kaggle/input/datasets/jashanpreetsingh645/agricultural-production-statistics/agricultural-production-statistics-jun02-final-tables-pigs-herd-size-territorial/3-pigs-territorial.xls",
    # Add remaining file paths here...
]
```

```python
dfs = {}

for file in file_paths:
    try:
        file_name = os.path.basename(file)

        if file.endswith(".xlsx"):
            df = pd.read_excel(file, engine="openpyxl")
        else:
            df = pd.read_excel(file)

        dfs[file_name] = df

        print(f"Loaded: {file_name}")
        print("Shape:", df.shape)
        print("-" * 50)

    except Exception as e:
        print(f"Error loading {file}: {e}")
```

```python
for name, df in dfs.items():
    print(f"\n{name}")
    display(df.head())
```

```python
combined_df = pd.concat(
    dfs.values(),
    ignore_index=True,
    sort=False
)

print("Combined Shape:", combined_df.shape)

combined_df.head()
```

```python
summary = pd.DataFrame([
    {
        "file_name": name,
        "rows": df.shape[0],
        "columns": df.shape[1]
    }
    for name, df in dfs.items()
])

summary.sort_values("rows", ascending=False)
```

```python
all_files = []

for root, dirs, files in os.walk(
    "/kaggle/input/datasets/jashanpreetsingh645/agricultural-production-statistics"
):
    for file in files:
        if file.endswith((".xls", ".xlsx")):
            all_files.append(os.path.join(root, file))

print("Total Excel Files:", len(all_files))
```

```python
dfs = {}

for file in all_files:
    try:
        file_name = os.path.basename(file)

        if file.endswith(".xlsx"):
            df = pd.read_excel(file, engine="openpyxl")
        else:
            df = pd.read_excel(file)

        dfs[file_name] = df

    except Exception as e:
        print(f"Error: {file_name} -> {e}")

print(f"Successfully loaded {len(dfs)} files")
```

## EDA

```python
sample_df = list(dfs.values())[0]

plt.figure(figsize=(14,6))
sns.heatmap(sample_df.isnull(), cbar=False)

plt.title("Missing Values Heatmap")
plt.show()
```

```python
dtype_df = pd.DataFrame(
    sample_df.dtypes,
    columns=['Datatype']
)

dtype_df['Count'] = 1

dtype_df.groupby('Datatype')['Count'].sum().plot(
    kind='bar',
    figsize=(10,5)
)

plt.title("Datatype Distribution")
plt.show()
```

```python
num_cols = sample_df.select_dtypes(
    include=['int64','float64']
).columns

print(num_cols)
```

```python
print(sample_df.dtypes)
```

```python
sample_df.info()
```

```python
plt.figure(figsize=(18,8))

sns.heatmap(
    sample_df.isnull(),
    cmap="viridis",
    cbar=True
)

plt.title(
    "Missing Values Heatmap",
    fontsize=18,
    color="darkred"
)

plt.show()
```

```python
plt.figure(figsize=(18,8))

sns.heatmap(
    sample_df.isnull(),
    cmap="viridis",
    cbar=True
)

plt.title(
    "Missing Values Heatmap",
    fontsize=18,
    color="darkred"
)

plt.show()
```

```python
missing_percent = (
    sample_df.isnull().sum() /
    len(sample_df)
) * 100

plt.figure(figsize=(16,6))

missing_percent.sort_values(
    ascending=False
).plot(
    kind='bar',
    color='crimson'
)

plt.title(
    "Missing Values Percentage",
    fontsize=18
)

plt.ylabel("% Missing")

plt.show()
```

```python
completeness = (
    sample_df.notnull().sum() /
    len(sample_df)
) * 100

plt.figure(figsize=(16,6))

completeness.plot(
    kind='bar',
    color='forestgreen'
)

plt.title(
    "Data Completeness %",
    fontsize=18
)

plt.ylabel("Completeness %")

plt.show()
```

```python
unique_counts = sample_df.nunique()

plt.figure(figsize=(16,6))

unique_counts.plot(
    kind='bar',
    color='darkorange'
)

plt.title(
    "Unique Values Per Column",
    fontsize=18
)

plt.ylabel("Unique Count")

plt.show()
```

```python
text_lengths = pd.DataFrame()

for col in sample_df.columns:

    text_lengths[col] = (
        sample_df[col]
        .astype(str)
        .str.len()
    )

plt.figure(figsize=(15,8))

sns.boxplot(
    data=text_lengths,
    palette="husl"
)

plt.xticks(rotation=90)

plt.title(
    "Text Length Distribution",
    fontsize=18
)

plt.show()
```

```python
plt.figure(figsize=(16,10))

sns.heatmap(
    sample_df.isnull().corr(),
    cmap='coolwarm',
    annot=True
)

plt.title(
    "Missing Value Correlation",
    fontsize=18
)

plt.show()
```

```python
occupancy = pd.DataFrame({
    'Column': sample_df.columns,
    'Filled': sample_df.notnull().sum()
})

plt.figure(figsize=(16,8))

sns.barplot(
    data=occupancy,
    x='Column',
    y='Filled',
    palette='Spectral'
)

plt.xticks(rotation=90)

plt.title(
    "Filled Records Per Column",
    fontsize=18
)

plt.show()
```

```python
filled = sample_df.notnull().sum().sum()
missing = sample_df.isnull().sum().sum()

plt.figure(figsize=(8,8))

plt.pie(
    [filled, missing],
    labels=['Filled', 'Missing'],
    colors=['limegreen','red'],
    autopct='%1.1f%%',
    explode=[0.05,0.05],
    shadow=True
)

plt.title(
    "Dataset Completeness",
    fontsize=18
)

plt.show()
```

```python
fig, ax = plt.subplots(
    2,
    2,
    figsize=(18,12)
)

sample_df.notnull().sum().plot(
    kind='bar',
    ax=ax[0,0],
    color='dodgerblue'
)

ax[0,0].set_title("Non Null Count")

sample_df.isnull().sum().plot(
    kind='bar',
    ax=ax[0,1],
    color='red'
)

ax[0,1].set_title("Missing Count")

sample_df.nunique().plot(
    kind='bar',
    ax=ax[1,0],
    color='orange'
)

ax[1,0].set_title("Unique Values")

(
    sample_df.notnull().sum()/len(sample_df)*100
).plot(
    kind='bar',
    ax=ax[1,1],
    color='green'
)

ax[1,1].set_title("Completeness %")

plt.tight_layout()
plt.show()
```
