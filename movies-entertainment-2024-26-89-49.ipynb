```python


# Display basic information
print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns)

print("\nFirst 5 Rows:")
display(df.head())

print("\nDataset Information:")
df.info()

# -----------------------------
# 3. Basic Statistics
# -----------------------------
print("\nStatistical Summary:")
display(df.describe(include="all").T)


# ============================================================
# 4. MISSING VALUES VISUALIZATION
# ============================================================

plt.figure(figsize=(14, 6))

missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

if len(missing) > 0:
    sns.barplot(
        x=missing.values,
        y=missing.index,
        palette="magma"
    )

    plt.title(
        "Missing Values by Column",
        fontsize=18,
        fontweight="bold"
    )
    plt.xlabel("Number of Missing Values")
    plt.ylabel("Columns")
    plt.tight_layout()
    plt.show()
else:
    print("No missing values found.")


# ============================================================
# 5. NUMERICAL COLUMNS
# ============================================================

numeric_cols = [
    "release_year",
    "release_month",
    "release_day",
    "rating_out_of_10",
    "vote_count",
    "runtime_minutes",
    "box_office_usd",
    "budget_usd"
]

# Keep only columns that actually exist
numeric_cols = [col for col in numeric_cols if col in df.columns]

print("\nNumerical Columns:")
print(numeric_cols)


# ============================================================
# 6. HISTOGRAMS FOR NUMERICAL FEATURES
# ============================================================

for col in numeric_cols:

    plt.figure(figsize=(10, 5))

    sns.histplot(
        data=df,
        x=col,
        kde=True,
        color="purple",
        edgecolor="black"
    )

    plt.title(
        f"Distribution of {col}",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel(col)
    plt.ylabel("Frequency")

    plt.grid(
        axis="y",
        linestyle="--",
        alpha=0.3
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 7. BOX PLOTS FOR OUTLIER DETECTION
# ============================================================

for col in numeric_cols:

    plt.figure(figsize=(10, 4))

    sns.boxplot(
        x=df[col],
        color="orange"
    )

    plt.title(
        f"Box Plot - {col}",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel(col)

    plt.grid(
        axis="x",
        linestyle="--",
        alpha=0.3
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 8. RELEASE YEAR DISTRIBUTION
# ============================================================

if "release_year" in df.columns:

    year_counts = df["release_year"].value_counts().sort_index()

    plt.figure(figsize=(15, 6))

    plt.plot(
        year_counts.index,
        year_counts.values,
        marker="o",
        linewidth=2,
        color="darkblue"
    )

    plt.title(
        "Number of Movies Released by Year",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Release Year")
    plt.ylabel("Number of Movies")

    plt.grid(
        True,
        linestyle="--",
        alpha=0.3
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 9. RELEASE MONTH DISTRIBUTION
# ============================================================

if "release_month" in df.columns:

    month_counts = df["release_month"].value_counts().sort_index()

    plt.figure(figsize=(12, 6))

    sns.barplot(
        x=month_counts.index,
        y=month_counts.values,
        palette="viridis"
    )

    plt.title(
        "Movies Released by Month",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Release Month")
    plt.ylabel("Number of Movies")

    plt.tight_layout()
    plt.show()


# ============================================================
# 10. RELEASE DAY DISTRIBUTION
# ============================================================

if "release_day" in df.columns:

    day_counts = df["release_day"].value_counts().sort_index()

    plt.figure(figsize=(14, 6))

    sns.barplot(
        x=day_counts.index,
        y=day_counts.values,
        palette="coolwarm"
    )

    plt.title(
        "Movies Released by Day of Month",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Release Day")
    plt.ylabel("Number of Movies")

    plt.tight_layout()
    plt.show()


# ============================================================
# 11. MOVIE RATING DISTRIBUTION
# ============================================================

if "rating_out_of_10" in df.columns:

    plt.figure(figsize=(10, 6))

    sns.histplot(
        df["rating_out_of_10"].dropna(),
        bins=20,
        kde=True,
        color="crimson",
        edgecolor="black"
    )

    plt.axvline(
        df["rating_out_of_10"].mean(),
        color="black",
        linestyle="--",
        linewidth=2,
        label="Mean Rating"
    )

    plt.title(
        "Movie Rating Distribution",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Rating out of 10")
    plt.ylabel("Number of Movies")

    plt.legend()

    plt.tight_layout()
    plt.show()


# ============================================================
# 12. CONTENT RATING COUNT
# ============================================================

if "content_rating" in df.columns:

    plt.figure(figsize=(12, 6))

    order = df["content_rating"].value_counts().index

    sns.countplot(
        data=df,
        y="content_rating",
        order=order,
        palette="Set2"
    )

    plt.title(
        "Movies by Content Rating",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Number of Movies")
    plt.ylabel("Content Rating")

    plt.tight_layout()
    plt.show()


# ============================================================
# 13. LANGUAGE DISTRIBUTION - TOP 15
# ============================================================

if "language" in df.columns:

    top_languages = (
        df["language"]
        .value_counts()
        .head(15)
    )

    plt.figure(figsize=(12, 7))

    sns.barplot(
        x=top_languages.values,
        y=top_languages.index,
        palette="plasma"
    )

    plt.title(
        "Top 15 Movie Languages",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Number of Movies")
    plt.ylabel("Language")

    plt.tight_layout()
    plt.show()


# ============================================================
# 14. COUNTRY DISTRIBUTION - TOP 15
# ============================================================

if "country" in df.columns:

    top_countries = (
        df["country"]
        .value_counts()
        .head(15)
    )

    plt.figure(figsize=(12, 7))

    sns.barplot(
        x=top_countries.values,
        y=top_countries.index,
        palette="cividis"
    )

    plt.title(
        "Top 15 Movie-Producing Countries",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Number of Movies")
    plt.ylabel("Country")

    plt.tight_layout()
    plt.show()


# ============================================================
# 15. TOP DIRECTORS
# ============================================================

if "director" in df.columns:

    top_directors = (
        df["director"]
        .dropna()
        .value_counts()
        .head(15)
    )

    plt.figure(figsize=(12, 7))

    sns.barplot(
        x=top_directors.values,
        y=top_directors.index,
        palette="rocket"
    )

    plt.title(
        "Top 15 Directors by Number of Movies",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Number of Movies")
    plt.ylabel("Director")

    plt.tight_layout()
    plt.show()


# ============================================================
# 16. TOP PRODUCTION COMPANIES
# ============================================================

if "production_company" in df.columns:

    top_companies = (
        df["production_company"]
        .dropna()
        .value_counts()
        .head(15)
    )

    plt.figure(figsize=(12, 7))

    sns.barplot(
        x=top_companies.values,
        y=top_companies.index,
        palette="mako"
    )

    plt.title(
        "Top 15 Production Companies",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Number of Movies")
    plt.ylabel("Production Company")

    plt.tight_layout()
    plt.show()


# ============================================================
# 17. TOP GENRES
# ============================================================

if "genres" in df.columns:

    # Split multiple genres
    genre_series = (
        df["genres"]
        .dropna()
        .astype(str)
        .str.split(",")
        .explode()
        .str.strip()
    )

    top_genres = genre_series.value_counts().head(15)

    plt.figure(figsize=(12, 7))

    sns.barplot(
        x=top_genres.values,
        y=top_genres.index,
        palette="Spectral"
    )

    plt.title(
        "Top 15 Movie Genres",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Number of Movies")
    plt.ylabel("Genre")

    plt.tight_layout()
    plt.show()


# ============================================================
# 18. GENRE PIE CHART
# ============================================================

if "genres" in df.columns:

    top_genres = genre_series.value_counts().head(8)

    plt.figure(figsize=(9, 9))

    plt.pie(
        top_genres.values,
        labels=top_genres.index,
        autopct="%1.1f%%",
        startangle=90,
        colors=sns.color_palette("tab10", len(top_genres))
    )

    plt.title(
        "Top Movie Genres - Percentage Distribution",
        fontsize=18,
        fontweight="bold"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 19. BUDGET VS BOX OFFICE
# ============================================================

if "budget_usd" in df.columns and "box_office_usd" in df.columns:

    plot_df = df[
        ["budget_usd", "box_office_usd"]
    ].dropna()

    plt.figure(figsize=(11, 7))

    sns.scatterplot(
        data=plot_df,
        x="budget_usd",
        y="box_office_usd",
        color="teal",
        alpha=0.6,
        s=70
    )

    plt.title(
        "Movie Budget vs Box Office Revenue",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Budget (USD)")
    plt.ylabel("Box Office (USD)")

    plt.grid(
        linestyle="--",
        alpha=0.3
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 20. RATING VS VOTE COUNT
# ============================================================

if "rating_out_of_10" in df.columns and "vote_count" in df.columns:

    plot_df = df[
        ["rating_out_of_10", "vote_count"]
    ].dropna()

    plt.figure(figsize=(11, 7))

    sns.scatterplot(
        data=plot_df,
        x="vote_count",
        y="rating_out_of_10",
        color="magenta",
        alpha=0.6,
        s=70
    )

    plt.title(
        "Vote Count vs Movie Rating",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Vote Count")
    plt.ylabel("Rating out of 10")

    plt.grid(
        linestyle="--",
        alpha=0.3
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 21. RUNTIME VS RATING
# ============================================================

if "runtime_minutes" in df.columns and "rating_out_of_10" in df.columns:

    plot_df = df[
        ["runtime_minutes", "rating_out_of_10"]
    ].dropna()

    plt.figure(figsize=(11, 7))

    sns.scatterplot(
        data=plot_df,
        x="runtime_minutes",
        y="rating_out_of_10",
        color="darkgreen",
        alpha=0.6,
        s=70
    )

    plt.title(
        "Runtime vs Movie Rating",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Runtime (Minutes)")
    plt.ylabel("Rating out of 10")

    plt.grid(
        linestyle="--",
        alpha=0.3
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 22. BUDGET DISTRIBUTION
# ============================================================

if "budget_usd" in df.columns:

    plt.figure(figsize=(11, 6))

    sns.histplot(
        df["budget_usd"].dropna(),
        bins=30,
        kde=True,
        color="gold",
        edgecolor="black"
    )

    plt.title(
        "Movie Budget Distribution",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Budget (USD)")
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()


# ============================================================
# 23. BOX OFFICE DISTRIBUTION
# ============================================================

if "box_office_usd" in df.columns:

    plt.figure(figsize=(11, 6))

    sns.histplot(
        df["box_office_usd"].dropna(),
        bins=30,
        kde=True,
        color="royalblue",
        edgecolor="black"
    )

    plt.title(
        "Box Office Revenue Distribution",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Box Office Revenue (USD)")
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()


# ============================================================
# 24. RUNTIME DISTRIBUTION
# ============================================================

if "runtime_minutes" in df.columns:

    plt.figure(figsize=(11, 6))

    sns.histplot(
        df["runtime_minutes"].dropna(),
        bins=30,
        kde=True,
        color="coral",
        edgecolor="black"
    )

    plt.title(
        "Movie Runtime Distribution",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Runtime (Minutes)")
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()


# ============================================================
# 25. TOP 15 MOVIES BY RATING
# ============================================================

if "title" in df.columns and "rating_out_of_10" in df.columns:

    top_movies_rating = (
        df[
            ["title", "rating_out_of_10"]
        ]
        .dropna()
        .sort_values(
            "rating_out_of_10",
            ascending=False
        )
        .head(15)
    )

    plt.figure(figsize=(12, 8))

    sns.barplot(
        data=top_movies_rating,
        x="rating_out_of_10",
        y="title",
        palette="flare"
    )

    plt.title(
        "Top 15 Movies by Rating",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Rating")
    plt.ylabel("Movie Title")

    plt.tight_layout()
    plt.show()


# ============================================================
# 26. TOP 15 MOVIES BY BOX OFFICE
# ============================================================

if "title" in df.columns and "box_office_usd" in df.columns:

    top_movies_boxoffice = (
        df[
            ["title", "box_office_usd"]
        ]
        .dropna()
        .sort_values(
            "box_office_usd",
            ascending=False
        )
        .head(15)
    )

    plt.figure(figsize=(12, 8))

    sns.barplot(
        data=top_movies_boxoffice,
        x="box_office_usd",
        y="title",
        palette="YlGnBu"
    )

    plt.title(
        "Top 15 Movies by Box Office Revenue",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Box Office Revenue (USD)")
    plt.ylabel("Movie Title")

    plt.tight_layout()
    plt.show()


# ============================================================
# 27. TOP 15 MOVIES BY VOTE COUNT
# ============================================================

if "title" in df.columns and "vote_count" in df.columns:

    top_voted_movies = (
        df[
            ["title", "vote_count"]
        ]
        .dropna()
        .sort_values(
            "vote_count",
            ascending=False
        )
        .head(15)
    )

    plt.figure(figsize=(12, 8))

    sns.barplot(
        data=top_voted_movies,
        x="vote_count",
        y="title",
        palette="GnBu"
    )

    plt.title(
        "Top 15 Movies by Vote Count",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Vote Count")
    plt.ylabel("Movie Title")

    plt.tight_layout()
    plt.show()


# ============================================================
# 28. CORRELATION HEATMAP
# ============================================================

numeric_corr_cols = df.select_dtypes(
    include=["int64", "int32", "float64", "float32"]
).columns.tolist()

print("Numeric columns used for correlation:")
print(numeric_corr_cols)

# Calculate correlation
correlation = df[numeric_corr_cols].corr()

# Plot heatmap
plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    center=0
)

plt.title(
    "Correlation Heatmap - Numerical Features",
    fontsize=18,
    fontweight="bold"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.yticks(
    rotation=0
)

plt.tight_layout()
plt.show()


# ============================================================
# 29. PAIRPLOT
# ============================================================

pair_cols = [
    "rating_out_of_10",
    "vote_count",
    "runtime_minutes",
    "budget_usd",
    "box_office_usd"
]

pair_cols = [
    col for col in pair_cols
    if col in df.columns
]

if len(pair_cols) >= 2:

    sns.pairplot(
        df[pair_cols].dropna().sample(
            min(1000, len(df[pair_cols].dropna())),
            random_state=42
        ),
        diag_kind="kde",
        palette="husl"
    )

    plt.suptitle(
        "Pairwise Relationships Between Numerical Features",
        y=1.02,
        fontsize=18,
        fontweight="bold"
    )

    plt.show()


# ============================================================
# 30. AVERAGE RATING BY YEAR
# ============================================================

if "release_year" in df.columns and "rating_out_of_10" in df.columns:

    yearly_rating = (
        df.groupby("release_year")["rating_out_of_10"]
        .mean()
        .dropna()
    )

    plt.figure(figsize=(15, 6))

    plt.plot(
        yearly_rating.index,
        yearly_rating.values,
        marker="o",
        linewidth=2,
        color="darkred"
    )

    plt.title(
        "Average Movie Rating by Release Year",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Release Year")
    plt.ylabel("Average Rating")

    plt.grid(
        linestyle="--",
        alpha=0.3
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 31. AVERAGE BOX OFFICE BY YEAR
# ============================================================

if "release_year" in df.columns and "box_office_usd" in df.columns:

    yearly_boxoffice = (
        df.groupby("release_year")["box_office_usd"]
        .mean()
        .dropna()
    )

    plt.figure(figsize=(15, 6))

    plt.plot(
        yearly_boxoffice.index,
        yearly_boxoffice.values,
        marker="o",
        linewidth=2,
        color="darkgreen"
    )

    plt.title(
        "Average Box Office Revenue by Release Year",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Release Year")
    plt.ylabel("Average Box Office Revenue (USD)")

    plt.grid(
        linestyle="--",
        alpha=0.3
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 32. AVERAGE RATING BY CONTENT RATING
# ============================================================

if "content_rating" in df.columns and "rating_out_of_10" in df.columns:

    content_rating_avg = (
        df.groupby("content_rating")["rating_out_of_10"]
        .mean()
        .sort_values(ascending=False)
    )

    plt.figure(figsize=(12, 6))

    sns.barplot(
        x=content_rating_avg.index,
        y=content_rating_avg.values,
        palette="Set1"
    )

    plt.title(
        "Average Movie Rating by Content Rating",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Content Rating")
    plt.ylabel("Average Rating")

    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()


# ============================================================
# 33. MOVIE COUNT BY YEAR - BAR CHART
# ============================================================

if "release_year" in df.columns:

    year_counts = (
        df["release_year"]
        .value_counts()
        .sort_index()
    )

    plt.figure(figsize=(16, 7))

    sns.barplot(
        x=year_counts.index,
        y=year_counts.values,
        color="slateblue"
    )

    plt.title(
        "Movie Count by Release Year",
        fontsize=18,
        fontweight="bold"
    )

    plt.xlabel("Release Year")
    plt.ylabel("Number of Movies")

    plt.xticks(rotation=90)

    plt.tight_layout()
    plt.show()


# ============================================================
# 34. FINAL EDA SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("EDA SUMMARY")
print("=" * 60)

print("Dataset Shape:", df.shape)

print("\nNumerical Columns:")
print(numeric_cols)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

if "rating_out_of_10" in df.columns:
    print(
        "\nAverage Movie Rating:",
        round(df["rating_out_of_10"].mean(), 2)
    )

if "runtime_minutes" in df.columns:
    print(
        "Average Runtime:",
        round(df["runtime_minutes"].mean(), 2),
        "minutes"
    )

if "budget_usd" in df.columns:
    print(
        "Average Budget:",
        round(df["budget_usd"].mean(), 2)
    )

if "box_office_usd" in df.columns:
    print(
        "Average Box Office:",
        round(df["box_office_usd"].mean(), 2)
    )

print("\nEDA Visualization Completed Successfully!")
```

## Feature engineering

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Machine Learning Algorithms
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor,
    ExtraTreesRegressor,
    RandomForestRegressor,
    HistGradientBoostingRegressor
)

from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR

print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nFirst 5 Records:")
display(df.head())


# ============================================================
# 3. DATASET INFORMATION
# ============================================================

print("\nDataset Information:")
df.info()


# ============================================================
# 4. CHECK MISSING VALUES
# ============================================================

print("\nMissing Values:")
print(df.isnull().sum())


# ============================================================
# 5. REMOVE DUPLICATES
# ============================================================

print("\nDuplicate Rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("Shape After Removing Duplicates:", df.shape)


# ============================================================
# 6. DEFINE TARGET
# ============================================================

TARGET = "rating_out_of_10"

print("\nTarget Variable:", TARGET)

print("\nTarget Statistics:")
print(df[TARGET].describe())


# ============================================================
# 7. REMOVE ROWS WHERE TARGET IS MISSING
# ============================================================

df = df.dropna(subset=[TARGET])

print("\nShape After Removing Missing Target:", df.shape)


# ============================================================
# 8. SELECT FEATURES
# ============================================================

# We remove:
# title, synopsis, URLs because they are text/identifiers
# rating_out_of_10 because it is our target

features = [
    "release_year",
    "release_month",
    "release_day",
    "vote_count",
    "runtime_minutes",
    "box_office_usd",
    "budget_usd",
    "director",
    "screenwriter",
    "cast",
    "production_company",
    "content_rating",
    "language",
    "country",
    "genres"
]

# Keep only columns available in dataset
features = [col for col in features if col in df.columns]

print("\nSelected Features:")
print(features)


# ============================================================
# 9. CREATE X AND y
# ============================================================

X = df[features].copy()

y = df[TARGET].copy()

print("\nX Shape:", X.shape)
print("y Shape:", y.shape)


# ============================================================
# 10. HANDLE CATEGORICAL AND NUMERICAL COLUMNS
# ============================================================

categorical_cols = X.select_dtypes(
    include=["object", "category"]
).columns.tolist()

numerical_cols = X.select_dtypes(
    include=["int64", "float64", "int32", "float32"]
).columns.tolist()

print("\nCategorical Columns:")
print(categorical_cols)

print("\nNumerical Columns:")
print(numerical_cols)


# ============================================================
# 11. LABEL ENCODING
# ============================================================

# LabelEncoder converts categorical values into numbers.
#
# Example:
# Action       -> 0
# Comedy       -> 1
# Drama        -> 2
#
# Each column receives its own encoder.

label_encoders = {}

for col in categorical_cols:

    # Convert missing values to string
    X[col] = X[col].fillna("Unknown").astype(str)

    encoder = LabelEncoder()

    X[col] = encoder.fit_transform(X[col])

    label_encoders[col] = encoder

    print(
        f"\nLabel Encoding: {col}"
    )

    print(
        dict(
            zip(
                encoder.classes_[:10],
                encoder.transform(encoder.classes_[:10])
            )
        )
    )


# ============================================================
# 12. HANDLE MISSING NUMERICAL VALUES
# ============================================================

for col in numerical_cols:

    X[col] = X[col].fillna(
        X[col].median()
    )


# ============================================================
# 13. CHECK X AFTER ENCODING
# ============================================================

print("\nX After Label Encoding:")
display(X.head())

print("\nX Data Types:")
print(X.dtypes)


# ============================================================
# 14. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTrain-Test Split")
print("----------------------------")

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)


# ============================================================
# 15. FEATURE SCALING
# ============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("\nFeature Scaling Completed.")


# ============================================================
# 16. CREATE ML MODELS
# ============================================================

models = {

    "Linear Regression":
        LinearRegression(),

    "Decision Tree":
        DecisionTreeRegressor(
            random_state=42,
            max_depth=10
        ),

    "Random Forest":
        RandomForestRegressor(
            n_estimators=200,
            random_state=42,
            n_jobs=-1,
            max_depth=15
        ),

    "Gradient Boosting":
        GradientBoostingRegressor(
            n_estimators=200,
            learning_rate=0.05,
            max_depth=5,
            random_state=42
        ),

    "Extra Trees":
        ExtraTreesRegressor(
            n_estimators=200,
            random_state=42,
            n_jobs=-1,
            max_depth=15
        ),

    "KNN":
        KNeighborsRegressor(
            n_neighbors=5
        ),

    "SVR":
        SVR(
            kernel="rbf",
            C=10,
            epsilon=0.1
        ),

    "Hist Gradient Boosting":
        HistGradientBoostingRegressor(
            max_iter=200,
            learning_rate=0.05,
            max_leaf_nodes=31,
            random_state=42
        )
}


# ============================================================
# 17. TRAIN ALL MODELS
# ============================================================

results = []

predictions = {}

for name, model in models.items():

    print("\n" + "=" * 60)
    print("Training:", name)
    print("=" * 60)

    # Train model
    model.fit(
        X_train_scaled,
        y_train
    )

    # Predict
    y_pred = model.predict(
        X_test_scaled
    )

    predictions[name] = y_pred

    # --------------------------------------------------------
    # MAE
    # --------------------------------------------------------

    mae = mean_absolute_error(
        y_test,
        y_pred
    )

    # --------------------------------------------------------
    # RMSE
    # --------------------------------------------------------

    rmse = np.sqrt(
        mean_squared_error(
            y_test,
            y_pred
        )
    )

    # --------------------------------------------------------
    # R2 SCORE
    # --------------------------------------------------------

    r2 = r2_score(
        y_test,
        y_pred
    )

    # --------------------------------------------------------
    # MAPE
    # --------------------------------------------------------

    # Avoid division by zero
    non_zero = y_test != 0

    if non_zero.sum() > 0:

        mape = np.mean(
            np.abs(
                (
                    y_test[non_zero]
                    -
                    y_pred[non_zero]
                )
                /
                y_test[non_zero]
            )
        ) * 100

    else:

        mape = np.nan


    # --------------------------------------------------------
    # ACCURACY %
    # --------------------------------------------------------

    accuracy = 100 - mape

    # Keep accuracy between 0 and 100
    accuracy = max(
        0,
        min(
            100,
            accuracy
        )
    )

    results.append({

        "Model": name,

        "Accuracy (%)":
            accuracy,

        "R2 Score":
            r2,

        "MAE":
            mae,

        "RMSE":
            rmse,

        "MAPE (%)":
            mape
    })


# ============================================================
# 18. MODEL COMPARISON
# ============================================================

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Accuracy (%)",
    ascending=False
)

print("\n")
print("=" * 80)
print("MODEL PERFORMANCE COMPARISON")
print("=" * 80)

display(
    results_df.round(4)
)


# ============================================================
# 19. ACCURACY PERCENTAGE
# ============================================================

print("\nMODEL ACCURACY")
print("=" * 60)

for _, row in results_df.iterrows():

    print(
        f"{row['Model']:<25} "
        f"{row['Accuracy (%)']:.2f}%"
    )


# ============================================================
# 20. R2 SCORE PERCENTAGE
# ============================================================

print("\nR2 SCORE AS PERCENTAGE")
print("=" * 60)

for _, row in results_df.iterrows():

    r2_percentage = row["R2 Score"] * 100

    print(
        f"{row['Model']:<25} "
        f"{r2_percentage:.2f}%"
    )


# ============================================================
# 21. ACCURACY COMPARISON GRAPH
# ============================================================

plt.figure(figsize=(14, 7))

sns.barplot(
    data=results_df,
    x="Accuracy (%)",
    y="Model",
    palette="viridis"
)

plt.title(
    "Machine Learning Model Accuracy Comparison",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel("Accuracy (%)")
plt.ylabel("Machine Learning Model")

plt.xlim(
    0,
    100
)

plt.grid(
    axis="x",
    linestyle="--",
    alpha=0.3
)

plt.tight_layout()
plt.show()


# ============================================================
# 22. R2 SCORE COMPARISON
# ============================================================

plt.figure(figsize=(14, 7))

sns.barplot(
    data=results_df,
    x="R2 Score",
    y="Model",
    palette="plasma"
)

plt.title(
    "R² Score Comparison",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel("R² Score")
plt.ylabel("Machine Learning Model")

plt.grid(
    axis="x",
    linestyle="--",
    alpha=0.3
)

plt.tight_layout()
plt.show()


# ============================================================
# 23. RMSE COMPARISON
# ============================================================

plt.figure(figsize=(14, 7))

sns.barplot(
    data=results_df.sort_values("RMSE"),
    x="RMSE",
    y="Model",
    palette="magma"
)

plt.title(
    "RMSE Comparison",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel("RMSE - Lower is Better")
plt.ylabel("Machine Learning Model")

plt.grid(
    axis="x",
    linestyle="--",
    alpha=0.3
)

plt.tight_layout()
plt.show()


# ============================================================
# 24. MAE COMPARISON
# ============================================================

plt.figure(figsize=(14, 7))

sns.barplot(
    data=results_df.sort_values("MAE"),
    x="MAE",
    y="Model",
    palette="coolwarm"
)

plt.title(
    "MAE Comparison",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel("Mean Absolute Error - Lower is Better")
plt.ylabel("Machine Learning Model")

plt.grid(
    axis="x",
    linestyle="--",
    alpha=0.3
)

plt.tight_layout()
plt.show()


# ============================================================
# 25. BEST MODEL
# ============================================================

best_model_name = results_df.iloc[0]["Model"]

best_accuracy = results_df.iloc[0]["Accuracy (%)"]

best_r2 = results_df.iloc[0]["R2 Score"]

best_rmse = results_df.iloc[0]["RMSE"]

best_mae = results_df.iloc[0]["MAE"]


print("\n" + "=" * 70)
print("BEST MODEL")
print("=" * 70)

print(
    "Best Model       :",
    best_model_name
)

print(
    f"Accuracy         : {best_accuracy:.2f}%"
)

print(
    f"R2 Score         : {best_r2:.4f}"
)

print(
    f"R2 Percentage    : {best_r2 * 100:.2f}%"
)

print(
    f"MAE              : {best_mae:.4f}"
)

print(
    f"RMSE             : {best_rmse:.4f}"
)


# ============================================================
# 26. ACTUAL VS PREDICTED FOR BEST MODEL
# ============================================================

best_predictions = predictions[
    best_model_name
]

plt.figure(figsize=(10, 7))

plt.scatter(
    y_test,
    best_predictions,
    alpha=0.6,
    color="purple"
)

# Perfect prediction line
min_value = min(
    y_test.min(),
    best_predictions.min()
)

max_value = max(
    y_test.max(),
    best_predictions.max()
)

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    color="red",
    linestyle="--",
    linewidth=2
)

plt.title(
    f"Actual vs Predicted Ratings - {best_model_name}",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel("Actual Rating")
plt.ylabel("Predicted Rating")

plt.grid(
    linestyle="--",
    alpha=0.3
)

plt.tight_layout()
plt.show()


# ============================================================
# 27. RESIDUAL PLOT
# ============================================================

residuals = (
    y_test.values
    -
    best_predictions
)

plt.figure(figsize=(11, 6))

plt.scatter(
    best_predictions,
    residuals,
    alpha=0.6,
    color="darkorange"
)

plt.axhline(
    y=0,
    color="black",
    linestyle="--"
)

plt.title(
    f"Residual Plot - {best_model_name}",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel("Predicted Rating")
plt.ylabel("Residual")

plt.grid(
    linestyle="--",
    alpha=0.3
)

plt.tight_layout()
plt.show()


# ============================================================
# 28. PREDICTION TABLE
# ============================================================

prediction_df = pd.DataFrame({

    "Actual Rating":
        y_test.values,

    "Predicted Rating":
        best_predictions

})

prediction_df["Error"] = (
    prediction_df["Actual Rating"]
    -
    prediction_df["Predicted Rating"]
)

prediction_df["Absolute Error"] = (
    np.abs(
        prediction_df["Error"]
    )
)

print("\nPrediction Results:")
display(
    prediction_df.head(20).round(3)
)


# ============================================================
# 29. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("FINAL MACHINE LEARNING SUMMARY")
print("=" * 80)

print(
    f"Total Records       : {len(df)}"
)

print(
    f"Training Records    : {len(X_train)}"
)

print(
    f"Testing Records     : {len(X_test)}"
)

print(
    f"Number of Features  : {X.shape[1]}"
)

print(
    f"Best Algorithm      : {best_model_name}"
)

print(
    f"Accuracy            : {best_accuracy:.2f}%"
)

print(
    f"R2 Score            : {best_r2:.4f}"
)

print(
    f"R2 Percentage       : {best_r2 * 100:.2f}%"
)

print(
    f"MAE                 : {best_mae:.4f}"
)

print(
    f"RMSE                : {best_rmse:.4f}"
)

print("=" * 80)
```
