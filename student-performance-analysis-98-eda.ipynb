## Feature engineering

```python

# ==============================================================
# UNIVERSITY STUDENT PASS/FAIL PREDICTION - COMPLETE ML PIPELINE
# Train/Test Split + Encoding + 7 ML Models + ROC Curves
# Target: Pass/Fail
# ==============================================================

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler, label_binarize
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report,
    roc_curve, auc, roc_auc_score
)

# --------------------------------------------------------------
# 1. LOAD / PREPARE DATA
# --------------------------------------------------------------
# Ensure df is already loaded, for example:
# df = pd.read_csv('/kaggle/input/your-dataset/student_data.csv')

df = df.copy()
df.columns = df.columns.str.strip()

target = "Pass/Fail"

if target not in df.columns:
    raise ValueError(f"Target column '{target}' not found. Check df.columns.")

# Remove records without target labels and duplicate rows
df = df.dropna(subset=[target]).drop_duplicates().copy()

# Remove rows with blank target values
df = df[df[target].astype(str).str.strip() != ""].copy()

# --------------------------------------------------------------
# 2. FEATURE SELECTION
# --------------------------------------------------------------
# Exclude identifiers, names, grades and marks to reduce leakage.
# Marks and Grade may directly determine Pass/Fail.
# If your marks are measured BEFORE the prediction point and are
# legitimately available, you can reconsider including Marks(%).

drop_cols = [
    "University-ID",
    "Name",
    "Pass/Fail",
    "Grade",
    "Marks(%)"
]

X = df.drop(columns=[c for c in drop_cols if c in df.columns])
y_raw = df[target].astype(str).str.strip()

# Remove features that have no useful variation
X = X.loc[:, X.nunique(dropna=False) > 1]

if X.shape[1] == 0:
    raise ValueError("No usable feature columns remain.")

if y_raw.nunique() < 2:
    raise ValueError("The target must contain at least two classes.")

print("=" * 70)
print("DATASET OVERVIEW")
print("=" * 70)
print("Dataset shape:", df.shape)
print("Feature shape:", X.shape)
print("Features:", X.columns.tolist())
print("\nTarget distribution:")
print(y_raw.value_counts())

# --------------------------------------------------------------
# 3. LABEL ENCODING FOR TARGET
# --------------------------------------------------------------
# LabelEncoder converts target labels such as Pass/Fail to numbers.
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y_raw)

class_names = label_encoder.classes_
n_classes = len(class_names)

print("\nEncoded target mapping:")
for i, name in enumerate(class_names):
    print(f"{name} -> {i}")

# --------------------------------------------------------------
# 4. TRAIN-TEST SPLITTING
# --------------------------------------------------------------
# Stratification maintains class proportions in both sets.
class_counts = pd.Series(y).value_counts()

if class_counts.min() < 2:
    raise ValueError(
        "At least 2 records per target class are required. "
        "Collect more examples for classes with very few records."
    )

test_size = max(0.2, n_classes / len(y))
test_size = min(test_size, 0.4)

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=test_size,
    random_state=42,
    stratify=y
)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))
print("Training target distribution:")
print(pd.Series(label_encoder.inverse_transform(y_train)).value_counts())
print("\nTesting target distribution:")
print(pd.Series(label_encoder.inverse_transform(y_test)).value_counts())

# --------------------------------------------------------------
# 5. NUMERIC AND CATEGORICAL FEATURE PROCESSING
# --------------------------------------------------------------
numeric_features = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    exclude=["number"]
).columns.tolist()

# Numeric data: median imputation + scaling
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical data: most-frequent imputation + one-hot encoding
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(transformers=[
    ("numeric", numeric_transformer, numeric_features),
    ("categorical", categorical_transformer, categorical_features)
])

print("\nNumeric features:", numeric_features)
print("Categorical features:", categorical_features)

# --------------------------------------------------------------
# 6. DEFINE MACHINE LEARNING ALGORITHMS
# --------------------------------------------------------------
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000, class_weight="balanced"
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=8, random_state=42, class_weight="balanced"
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200, random_state=42,
        class_weight="balanced", n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100, learning_rate=0.1,
        max_depth=3, random_state=42
    ),

    "K-Nearest Neighbors": KNeighborsClassifier(
        n_neighbors=max(1, min(5, len(X_train)))
    ),

    "Support Vector Machine": SVC(
        probability=True, class_weight="balanced",
        random_state=42
    ),

    "Gaussian Naive Bayes": GaussianNB()
}

# --------------------------------------------------------------
# 7. TRAIN MODELS AND EVALUATE PERFORMANCE
# --------------------------------------------------------------
results = []
trained_models = {}
predictions = {}
probabilities = {}

for name, model in models.items():
    print("\n" + "=" * 70)
    print("TRAINING:", name)
    print("=" * 70)

    try:
        # GaussianNB requires dense input; other models accept sparse data.
        if name == "Gaussian Naive Bayes":
            from sklearn.preprocessing import FunctionTransformer
            dense_preprocessor = Pipeline(steps=[
                ("features", preprocessor),
                ("dense", FunctionTransformer(
                    lambda data: data.toarray()
                    if hasattr(data, "toarray") else data
                ))
            ])
            pipeline = Pipeline(steps=[
                ("preprocessing", dense_preprocessor),
                ("model", model)
            ])
        else:
            pipeline = Pipeline(steps=[
                ("preprocessing", preprocessor),
                ("model", model)
            ])

        pipeline.fit(X_train, y_train)
        y_pred = pipeline.predict(X_test)

        # Probability scores for ROC curves
        y_prob = pipeline.predict_proba(X_test)

        trained_models[name] = pipeline
        predictions[name] = y_pred
        probabilities[name] = y_prob

        accuracy = accuracy_score(y_test, y_pred)
        precision = precision_score(
            y_test, y_pred, average="weighted", zero_division=0
        )
        recall = recall_score(
            y_test, y_pred, average="weighted", zero_division=0
        )
        f1 = f1_score(
            y_test, y_pred, average="weighted", zero_division=0
        )

        # Macro ROC-AUC works for binary and multiclass classification.
        if n_classes == 2:
            auc_score = roc_auc_score(y_test, y_prob[:, 1])
        else:
            auc_score = roc_auc_score(
                y_test, y_prob,
                multi_class="ovr", average="macro", labels=np.arange(n_classes)
            )

        results.append({
            "Model": name,
            "Accuracy (%)": accuracy * 100,
            "Precision (%)": precision * 100,
            "Recall (%)": recall * 100,
            "F1 Score (%)": f1 * 100,
            "ROC-AUC": auc_score
        })

        print(f"Accuracy : {accuracy * 100:.2f}%")
        print(f"Precision: {precision * 100:.2f}%")
        print(f"Recall   : {recall * 100:.2f}%")
        print(f"F1 Score : {f1 * 100:.2f}%")
        print(f"ROC-AUC  : {auc_score:.4f}")

        print("\nClassification Report:")
        print(classification_report(
            y_test, y_pred,
            labels=np.arange(n_classes),
            target_names=class_names,
            zero_division=0
        ))

    except Exception as e:
        print(f"Model failed: {e}")

if not results:
    raise RuntimeError("No models trained successfully.")

# --------------------------------------------------------------
# 8. MODEL COMPARISON TABLE
# --------------------------------------------------------------
results_df = pd.DataFrame(results).sort_values(
    by="F1 Score (%)", ascending=False
).reset_index(drop=True)

print("\n" + "=" * 70)
print("MODEL COMPARISON")
print("=" * 70)
print(results_df.round(4).to_string(index=False))

# --------------------------------------------------------------
# 9. COMPARE ACCURACY, PRECISION, RECALL AND F1
# --------------------------------------------------------------
metric_cols = [
    "Accuracy (%)", "Precision (%)",
    "Recall (%)", "F1 Score (%)"
]

comparison = results_df.set_index("Model")[metric_cols]

ax = comparison.plot(
    kind="bar", figsize=(14, 7),
    colormap="Set2", edgecolor="black", width=0.8
)
ax.set_title("Machine Learning Model Performance Comparison")
ax.set_ylabel("Score (%)")
ax.set_xlabel("Machine Learning Algorithm")
ax.set_ylim(0, 105)
plt.xticks(rotation=30, ha="right")
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()

# --------------------------------------------------------------
# 10. ROC CURVES - BINARY CLASSIFICATION
# --------------------------------------------------------------
if n_classes == 2:
    plt.figure(figsize=(10, 8))

    for name, y_prob in probabilities.items():
        fpr, tpr, _ = roc_curve(y_test, y_prob[:, 1])
        roc_auc = auc(fpr, tpr)

        plt.plot(
            fpr, tpr, linewidth=2,
            label=f"{name} (AUC = {roc_auc:.3f})"
        )

    plt.plot([0, 1], [0, 1], "k--", label="Random Classifier")
    plt.title("ROC Curves - Pass/Fail Prediction")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.legend(loc="lower right", fontsize=8)
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

# --------------------------------------------------------------
# 11. ROC CURVES - MULTICLASS CLASSIFICATION
# --------------------------------------------------------------
else:
    y_test_binary = label_binarize(
        y_test, classes=np.arange(n_classes)
    )

    for name, y_prob in probabilities.items():
        plt.figure(figsize=(9, 7))

        for i, class_name in enumerate(class_names):
            fpr, tpr, _ = roc_curve(
                y_test_binary[:, i], y_prob[:, i]
            )
            class_auc = auc(fpr, tpr)

            plt.plot(
                fpr, tpr, linewidth=2,
                label=f"{class_name} (AUC = {class_auc:.3f})"
            )

        plt.plot([0, 1], [0, 1], "k--", label="Random Classifier")
        plt.title(f"Multiclass ROC Curve - {name}")
        plt.xlabel("False Positive Rate")
        plt.ylabel("True Positive Rate")
        plt.legend(loc="lower right", fontsize=8)
        plt.grid(alpha=0.3)
        plt.tight_layout()
        plt.show()

# --------------------------------------------------------------
# 12. CONFUSION MATRIX FOR EACH MODEL
# --------------------------------------------------------------
for name, y_pred in predictions.items():
    cm = confusion_matrix(
        y_test, y_pred, labels=np.arange(n_classes)
    )

    plt.figure(figsize=(7, 5))
    sns.heatmap(
        cm, annot=True, fmt="d", cmap="YlGnBu",
        xticklabels=class_names,
        yticklabels=class_names
    )
    plt.title(f"Confusion Matrix - {name}")
    plt.xlabel("Predicted Class")
    plt.ylabel("Actual Class")
    plt.tight_layout()
    plt.show()

# --------------------------------------------------------------
# 13. BEST MODEL BASED ON WEIGHTED F1 SCORE
# --------------------------------------------------------------
best_name = results_df.iloc[0]["Model"]
best_model = trained_models[best_name]

print("\n" + "=" * 70)
print("BEST MODEL")
print("=" * 70)
print("Selected by weighted F1 score:", best_name)
print(results_df.iloc[0].to_string())

# Detailed prediction comparison
best_predictions = predictions[best_name]

prediction_results = pd.DataFrame({
    "Actual Result": label_encoder.inverse_transform(y_test),
    "Predicted Result": label_encoder.inverse_transform(best_predictions),
    "Correct Prediction": y_test == best_predictions
})

print("\nSample predictions from the test set:")
print(prediction_results.head(20).to_string(index=False))

print("\nPrediction accuracy on test set:",
      f"{accuracy_score(y_test, best_predictions) * 100:.2f}%")

print("\nMachine learning training and evaluation completed!")
```
