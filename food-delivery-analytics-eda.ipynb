## Feature engineering


```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import (
    LabelEncoder,
    OneHotEncoder,
    StandardScaler,
    label_binarize
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier,
    ExtraTreesClassifier,
    AdaBoostClassifier
)
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    auc,
    roc_auc_score
)

from sklearn.base import clone

# Visualization settings
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

print("="*70)
print("FOOD DELIVERY MACHINE LEARNING CLASSIFICATION")
print("="*70)

print("\nDataset Shape:", df.shape)
print("\nFirst Five Records:")
display(df.head())

print("\nDataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

# ================================================================
# 3. DATA CLEANING
# ================================================================

df = df.drop_duplicates().copy()

# Target variable
target = 'Churn_Risk'

# Remove rows where target is missing
df = df.dropna(subset=[target]).copy()

# Remove identifier column
if 'Order_ID' in df.columns:
    df.drop(columns=['Order_ID'], inplace=True)

# Separate numerical and categorical columns
numeric_cols = df.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

categorical_cols = df.select_dtypes(
    include=['object', 'category', 'bool']
).columns.tolist()

if target in numeric_cols:
    numeric_cols.remove(target)

if target in categorical_cols:
    categorical_cols.remove(target)

print("\nNumerical Features:")
print(numeric_cols)

print("\nCategorical Features:")
print(categorical_cols)

print("\nTarget Distribution:")
print(df[target].value_counts())

# ================================================================
# 4. DEFINE X AND Y
# ================================================================

X = df.drop(columns=[target])
y = df[target].astype(str)

print("\nIndependent Variables Shape:", X.shape)
print("Dependent Variable Shape:", y.shape)

# ================================================================
# 5. LABEL ENCODING TARGET VARIABLE
# ================================================================

le = LabelEncoder()

y_encoded = le.fit_transform(y)

print("\nTarget Classes:")
for i, class_name in enumerate(le.classes_):
    print(i, "=", class_name)

print("\nEncoded Target Sample:")
print(y_encoded[:20])

# ================================================================
# 6. TRAIN TEST SPLITTING
# ================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("\nTraining Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)

print("\nTraining Target Distribution:")
print(pd.Series(y_train).value_counts())

print("\nTesting Target Distribution:")
print(pd.Series(y_test).value_counts())

# ================================================================
# 7. FEATURE ENCODING AND SCALING
# ================================================================

# Numerical preprocessing
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Categorical preprocessing
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(
        handle_unknown='ignore',
        sparse_output=False
    ))
])

# Combine both transformations
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_cols),
        ('cat', categorical_transformer, categorical_cols)
    ],
    remainder='drop'
)

# Fit only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform test data without fitting again
X_test_processed = preprocessor.transform(X_test)

print("\nAfter Encoding and Scaling:")
print("X Train:", X_train_processed.shape)
print("X Test:", X_test_processed.shape)

print("\nFeature preprocessing completed!")

# ================================================================
# 8. DEFINE MACHINE LEARNING ALGORITHMS
# ================================================================

models = {
    'Logistic Regression': LogisticRegression(
        max_iter=2000,
        class_weight='balanced',
        random_state=42
    ),

    'Decision Tree': DecisionTreeClassifier(
        max_depth=12,
        min_samples_split=5,
        class_weight='balanced',
        random_state=42
    ),

    'Random Forest': RandomForestClassifier(
        n_estimators=200,
        max_depth=15,
        min_samples_split=5,
        class_weight='balanced',
        n_jobs=-1,
        random_state=42
    ),

    'Extra Trees': ExtraTreesClassifier(
        n_estimators=200,
        max_depth=15,
        class_weight='balanced',
        n_jobs=-1,
        random_state=42
    ),

    'Gradient Boosting': GradientBoostingClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        random_state=42
    ),

    'AdaBoost': AdaBoostClassifier(
        n_estimators=100,
        learning_rate=0.5,
        random_state=42
    ),

    'KNN': KNeighborsClassifier(
        n_neighbors=5,
        n_jobs=-1
    ),

    'Support Vector Machine': SVC(
        kernel='rbf',
        probability=True,
        random_state=42
    ),

    'Gaussian Naive Bayes': GaussianNB()
}

# ================================================================
# 9. TRAIN ALL MODELS AND EVALUATE
# ================================================================

results = []
trained_models = {}
roc_results = {}

number_of_classes = len(np.unique(y_train))

for name, model in models.items():

    print("\n" + "="*70)
    print("TRAINING:", name)
    print("="*70)

    try:

        # Train model
        model.fit(X_train_processed, y_train)

        # Predictions
        y_pred = model.predict(X_test_processed)

        # Probability predictions
        y_prob = model.predict_proba(X_test_processed)

        # Accuracy
        accuracy = accuracy_score(y_test, y_pred) * 100

        # Precision
        precision = precision_score(
            y_test, y_pred,
            average='weighted',
            zero_division=0
        ) * 100

        # Recall
        recall = recall_score(
            y_test, y_pred,
            average='weighted',
            zero_division=0
        ) * 100

        # F1 Score
        f1 = f1_score(
            y_test, y_pred,
            average='weighted',
            zero_division=0
        ) * 100

        # ROC AUC
        if number_of_classes == 2:

            roc_auc = roc_auc_score(
                y_test,
                y_prob[:, 1]
            )

            fpr, tpr, _ = roc_curve(
                y_test,
                y_prob[:, 1]
            )

        else:

            y_test_binary = label_binarize(
                y_test,
                classes=np.arange(number_of_classes)
            )

            roc_auc = roc_auc_score(
                y_test_binary,
                y_prob,
                average='macro',
                multi_class='ovr'
            )

            fpr, tpr, _ = roc_curve(
                y_test_binary.ravel(),
                y_prob.ravel()
            )

        # Store ROC results
        roc_results[name] = {
            'fpr': fpr,
            'tpr': tpr,
            'auc': roc_auc
        }

        # Save trained models
        trained_models[name] = model

        # Save metrics
        results.append({
            'Model': name,
            'Accuracy (%)': accuracy,
            'Precision (%)': precision,
            'Recall (%)': recall,
            'F1 Score (%)': f1,
            'ROC AUC (%)': roc_auc * 100
        })

        print(f"\nAccuracy  : {accuracy:.2f}%")
        print(f"Precision : {precision:.2f}%")
        print(f"Recall    : {recall:.2f}%")
        print(f"F1 Score  : {f1:.2f}%")
        print(f"ROC AUC   : {roc_auc*100:.2f}%")

        print("\nClassification Report:")
        print(classification_report(
            y_test,
            y_pred,
            labels=np.arange(len(le.classes_)),
            target_names=le.classes_,
            zero_division=0
        ))

    except Exception as e:
        print("Model Failed:", name)
        print("Reason:", str(e))

# ================================================================
# 10. MODEL COMPARISON TABLE
# ================================================================

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by='Accuracy (%)',
    ascending=False
).reset_index(drop=True)

print("\n" + "="*70)
print("MACHINE LEARNING MODEL COMPARISON")
print("="*70)

display(
    results_df.style.format({
        'Accuracy (%)': '{:.2f}%',
        'Precision (%)': '{:.2f}%',
        'Recall (%)': '{:.2f}%',
        'F1 Score (%)': '{:.2f}%',
        'ROC AUC (%)': '{:.2f}%'
    }).background_gradient(cmap='YlGn')
)

# ================================================================
# 11. ACCURACY COMPARISON
# ================================================================

plt.figure(figsize=(15, 7))

sns.barplot(
    data=results_df,
    x='Model',
    y='Accuracy (%)',
    hue='Model',
    palette='viridis',
    legend=False
)

plt.title('Machine Learning Models - Accuracy Comparison',
          fontsize=16)

plt.xlabel('Machine Learning Algorithms')
plt.ylabel('Accuracy (%)')
plt.xticks(rotation=40, ha='right')
plt.ylim(0, 100)

for i, value in enumerate(results_df['Accuracy (%)']):
    plt.text(i, value + 0.5, f'{value:.2f}%',
             ha='center', fontsize=9)

plt.tight_layout()
plt.show()

# ================================================================
# 12. PRECISION, RECALL AND F1 SCORE COMPARISON
# ================================================================

metrics_df = results_df.melt(
    id_vars='Model',
    value_vars=[
        'Precision (%)',
        'Recall (%)',
        'F1 Score (%)'
    ],
    var_name='Metric',
    value_name='Score'
)

plt.figure(figsize=(15, 7))

sns.barplot(
    data=metrics_df,
    x='Model',
    y='Score',
    hue='Metric',
    palette='Set2'
)

plt.title('Precision, Recall and F1 Score Comparison',
          fontsize=16)

plt.ylabel('Score (%)')
plt.xticks(rotation=40, ha='right')
plt.ylim(0, 100)
plt.legend(title='Evaluation Metrics')

plt.tight_layout()
plt.show()

# ================================================================
# 13. ROC CURVE - ALL MACHINE LEARNING MODELS
# ================================================================

plt.figure(figsize=(12, 9))

colors = sns.color_palette(
    'husl',
    n_colors=len(roc_results)
)

for (name, data), color in zip(
    roc_results.items(), colors
):

    plt.plot(
        data['fpr'],
        data['tpr'],
        color=color,
        linewidth=2,
        label=f"{name} (AUC = {data['auc']:.3f})"
    )

plt.plot(
    [0, 1],
    [0, 1],
    linestyle='--',
    color='black',
    label='Random Classifier'
)

plt.title('ROC Curve - Machine Learning Model Comparison',
          fontsize=16)

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')

plt.legend(loc='lower right')
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

# ================================================================
# 14. CONFUSION MATRIX FOR EACH MODEL
# ================================================================

for name, model in trained_models.items():

    y_pred = model.predict(X_test_processed)

    cm = confusion_matrix(
        y_test,
        y_pred,
        labels=np.arange(len(le.classes_))
    )

    plt.figure(figsize=(8, 6))

    sns.heatmap(
        cm,
        annot=True,
        fmt='d',
        cmap='YlGnBu',
        xticklabels=le.classes_,
        yticklabels=le.classes_
    )

    plt.title(f'Confusion Matrix - {name}')
    plt.xlabel('Predicted Class')
    plt.ylabel('Actual Class')

    plt.tight_layout()
    plt.show()

# ================================================================
# 15. ROC AUC SCORE COMPARISON
# ================================================================

auc_df = results_df.sort_values(
    by='ROC AUC (%)',
    ascending=False
)

plt.figure(figsize=(14, 7))

sns.barplot(
    data=auc_df,
    x='Model',
    y='ROC AUC (%)',
    hue='Model',
    palette='magma',
    legend=False
)

plt.title('ROC AUC Score Comparison',
          fontsize=16)

plt.ylabel('ROC AUC (%)')
plt.xticks(rotation=40, ha='right')
plt.ylim(0, 100)

plt.tight_layout()
plt.show()

# ================================================================
# 16. BEST MODEL SELECTION
# ================================================================

if len(results_df) > 0:

    best_model_name = results_df.iloc[0]['Model']

    best_accuracy = results_df.iloc[0]['Accuracy (%)']

    best_model = trained_models[best_model_name]

    print("\n" + "="*70)
    print("                 BEST MACHINE LEARNING MODEL")
    print("="*70)

    print("\nBest Model:", best_model_name)
    print(f"Best Accuracy: {best_accuracy:.2f}%")

    print(
        f"Precision: {results_df.iloc[0]['Precision (%)']:.2f}%"
    )

    print(
        f"Recall: {results_df.iloc[0]['Recall (%)']:.2f}%"
    )

    print(
        f"F1 Score: {results_df.iloc[0]['F1 Score (%)']:.2f}%"
    )

    print(
        f"ROC AUC: {results_df.iloc[0]['ROC AUC (%)']:.2f}%"
    )

    print("\nBest Model Classification Report:")

    best_predictions = best_model.predict(
        X_test_processed
    )

    print(classification_report(
        y_test,
        best_predictions,
        target_names=le.classes_,
        zero_division=0
    ))

else:

    print("No models trained successfully.")

# ================================================================
# 17. FINAL SUMMARY
# ================================================================

print("\n" + "="*70)
print("             MACHINE LEARNING EXECUTION COMPLETED")
print("="*70)

print("\nTotal Models Trained:", len(trained_models))

if len(results_df) > 0:
    print("Best Algorithm:", best_model_name)
    print(f"Best Accuracy: {best_accuracy:.2f}%")

print("\nTarget Variable:", target)
print("Train-Test Split: 80:20")
print("Random State: 42")
print("Feature Scaling: StandardScaler")
print("Categorical Encoding: OneHotEncoder")
print("Target Encoding: LabelEncoder")

print("\nAll Models Evaluated Successfully!")
```
