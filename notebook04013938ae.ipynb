```python
@dataclass
class Settings:
    target: str
    identifier: str | None
    shuffle: bool
    splits: int
    seed:int
    path:str
    

    def __repr__(self) -> str:
        return f"""
        {"="*12} Settings {"="*12}
        ==> Target: {self.target}
        ==> Identifier: {self.identifier}
        ==> Shuffle: {self.shuffle}
        ==> Number of splits: {self.splits}
        ==> Seed: {self.seed}
        {"="*34}
        """

CFG = Settings(
    target="Will_Buy_EV",
    identifier="id",
    shuffle=True,
    splits=3,
    seed=152,
    path="/kaggle/input/competitions/playground-series-s6e9"
)

np.random.seed(CFG.seed)

CFG
```

```python
def read_data(directory:str) -> tuple:
    """
        Read CSV files from a directory and return train, test, and submission DataFrames.
    """
    data = {}
    for filename in os.listdir(directory):
        key = re.search("(.*).csv", filename).group(1)
        data[key] = pd.read_csv(
            os.path.join(
                directory,
                filename
            )
        )

    train = data["train"]
    test  = data["test"]
    sub   = data["sample_submission"]

    return train, test, sub
```

```python
def plot_kde(df1: pd.DataFrame,
             df2: pd.DataFrame,
             cols: int = 3,
             fill: bool = True,
             figsize: tuple = (15, 12),
             colors: list = ["#009E73", "#D55E00"],
             title: str = None) -> None:

    df1 = df1.copy()
    df2 = df2.copy()

    df1.drop([CFG.target], axis=1, inplace=True)

    numeric_columns = df1.select_dtypes(include='number').columns
    rows = np.ceil(len(numeric_columns) / cols).astype(int)

    fig, ax = plt.subplots(nrows=rows, ncols=cols, figsize=figsize)
    fig.patch.set_facecolor("#FFFCFC")

    for i, x in enumerate(numeric_columns):
        current_ax = ax[i // cols, i % cols] if rows > 1 else ax[i % cols]
        current_ax.set_facecolor("#FFFCFC")
        sns.kdeplot(x=x, color=colors[0], fill=fill, data=df1, edgecolor="black", label='Train', ax=current_ax, linewidth=2.5)
        sns.kdeplot(x=x, color=colors[1], fill=fill, data=df2, edgecolor="black", label='Test', ax=current_ax, linewidth=2.5)
        current_ax.set_xlabel(x, fontsize=14, weight="bold")
        current_ax.set_ylabel("Density", fontsize=12)
        current_ax.tick_params(labelsize=11)

        current_ax.spines["left"].set_visible(False)
        current_ax.spines["bottom"].set_visible(True)
        current_ax.spines["right"].set_visible(False)
        current_ax.spines["top"].set_visible(False)

        current_ax.spines["bottom"].set_color("#000")
        current_ax.spines["bottom"].set_linewidth(2)
        
        current_ax.xaxis.grid(False)
        current_ax.yaxis.grid(True, color="#000", linestyle="--", linewidth=1.0, alpha=0.2, zorder=0)
        current_ax.set_axisbelow(True)

    if len(numeric_columns) < (rows * cols):
        for i in range(len(numeric_columns), rows * cols):
            fig.delaxes(np.atleast_1d(ax).flatten()[i])

    legend_elements = [
        Patch(facecolor=colors[0], edgecolor='black', label='Train'),
        Patch(facecolor=colors[1], edgecolor='black', label='Test')
    ]

    fig.legend(handles=legend_elements, loc='upper center', ncol=2,
               bbox_to_anchor=(0.5, 0.97), frameon=False, fontsize=13)

    if title:
        fig.suptitle(title, fontsize=22, weight="bold", y=0.99)

    fig.subplots_adjust(left=0.07, right=0.97, top=0.90, bottom=0.07, hspace=0.5, wspace=0.3)
    plt.show()
```

```python
def validate_hypothesis(models, X, y, noise_cols, splits=5):
    X = X.copy()
    noise_cols = noise_cols.copy()
    
    results = defaultdict(list)

    console.print(f"[bold {ORANGE}]Baseline [/bold {ORANGE}]")
    for name, model in models.items():
        result = cross_val_score(model, X, y, cv=splits, scoring="roc_auc")
        results[name].append(result)
        print(f"{name}: {result.mean():.4f} ± {result.std():.4f}")

    while len(noise_cols) > 0:
        col = noise_cols.pop(0)
        X.drop(col, axis=1, inplace=True)

        console.print(f"\n[bold {STRONG_ORANGE}]Removed:[/bold {STRONG_ORANGE}] {col}")
        for name, model in models.items():
            result = cross_val_score(model, X, y, cv=splits, scoring="roc_auc")
            results[name].append(result)
            print(f"{name}: {result.mean():.4f} ± {result.std():.4f}")

    return results
```

```python
# Most probable to be just noise to less probable (noise as negative control)
candidates = ["noise", "Charging_Stations_Near_Work", "Charging_Stations_Near_Home", "Gender", "Current_Car_Type", "City_Type", "Home_Charging_Possible", "Age", "Range_Anxiety_Level"]

results = validate_hypothesis(models, X, y, candidates)
```

```python
def apply_engineered_features(df):
    df = df.copy()
    df["Income_Env_Ratio"] = df["Annual_Income_USD"]/(df["Environmental_Concern_Level"]+1e-5)
    df["Subsidy_Anxiety"] = ((df["Subsidy_Available"].astype(str) + "_" + df["Range_Anxiety_Level"].astype(str)) == "1_1").astype(int)
    df["Comute_Env_ratio"] = df["Daily_Commute_km"]/(df["Environmental_Concern_Level"]+1e-5)
    df["Age_plus_Annual_Income_USD"] = df["Age"] + df["Annual_Income_USD"]
    df["Age_minus_Number_of_Cars_Owned"] = df["Age"] - df["Number_of_Cars_Owned"]
    
    return df
```

```python
def cross_validation(model, X, y, test, cv, n_seeds=3, to_encode=None, eval_set=False, bin_income=False):
    oofs = {}
    all_scores = []

    if to_encode is not None:
        encoded = [f"{c}_te" for c in to_encode]

    for seed in range(n_seeds):
        current_seed = CFG.seed + seed
        oof_train_seed = np.zeros(X.shape[0])
        oof_test_seed = np.zeros(test.shape[0])
        seed_scores = []

        cv.random_state = current_seed

        for i, (train_idx, val_idx) in enumerate(cv.split(X, y)):
            X_train, y_train = X.iloc[train_idx].copy(), y.iloc[train_idx]
            X_val, y_val = X.iloc[val_idx].copy(), y.iloc[val_idx]
            test_fold = test.copy()

            if bin_income:
                kb = KBinsDiscretizer(n_bins=600, encode="ordinal", strategy="quantile")
                X_train["inc_bin_600"] = kb.fit_transform(X_train[["Annual_Income_USD"]]).ravel().astype("int32")
                X_val["inc_bin_600"] = kb.transform(X_val[["Annual_Income_USD"]]).ravel().astype("int32")
                test_fold["inc_bin_600"] = kb.transform(test_fold[["Annual_Income_USD"]]).ravel().astype("int32")

            if to_encode is not None:
                te = TargetEncoder(target_type="binary", smooth="auto")
                X_train[encoded] = te.fit_transform(X_train[to_encode].astype(str), y_train)
                X_val[encoded] = te.transform(X_val[to_encode].astype(str))
                test_fold[encoded] = te.transform(test_fold[to_encode].astype(str))

            clf = clone(model)
            clf.random_state = current_seed

            if eval_set:
                clf.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=False)
            else:
                clf.fit(X_train, y_train)

            p_val = clf.predict_proba(X_val)[:, 1]
            p_test = clf.predict_proba(test_fold)[:, 1]

            oof_train_seed[val_idx] = p_val
            oof_test_seed += p_test / CFG.splits

            score = roc_auc_score(y_val, p_val)
            seed_scores.append(score)
            all_scores.append(score)

            console.print(f"[bold] Seed {current_seed} | Fold {i + 1}: {score:.4f} [/bold]")

        console.print(f"[bold {ORANGE}]Seed {current_seed}: {np.mean(seed_scores):.4f} ± {np.std(seed_scores):.4f}[/]")
        print()

        oofs[f"seed_{current_seed}"] = {
            "train": oof_train_seed,
            "test": oof_test_seed,
            "scores": seed_scores,
            "mean_score": np.mean(seed_scores)
        }

    console.print(f"[bold {STRONG_ORANGE}]Overall: {np.mean(all_scores):.4f} ± {np.std(all_scores):.4f}[/]")

    return oofs
```

```python
skf =  StratifiedKFold(
    n_splits=CFG.splits,
    shuffle=True,
    random_state=CFG.seed
)

oofs = cross_validation(
    model, X, y, test, skf, 
    to_encode=["Annual_Income_USD", "Daily_Commute_km", "inc_bin_600", "income1000_floor",
               "income100_floor", "commute_integer", "inc_d0", "inc_d1", "inc_d2", "inc_d3",
               "Age", "Number_of_Cars_Owned", "Home_Charging_Possible",
               "Subsidy_Available", "Range_Anxiety_Level",
               "Gender", "Charging_Stations_Near_Work", "Charging_Stations_Near_Home",
               "Current_Car_Type", "City_Type"],
    eval_set=True, 
    bin_income=True
)
```

```python
oofs_train = pd.DataFrame({seed: data["train"] for seed, data in oofs.items()})
oofs_test = pd.DataFrame({seed: data["test"] for seed, data in oofs.items()})

calibrators = {}

for col in oofs_train.columns:
    raw = oofs_train[col].to_numpy()
    calibrated = np.zeros_like(raw)

    for train_idx, val_idx in skf.split(raw, y):
        iso = IsotonicRegression(out_of_bounds="clip")
        iso.fit(raw[train_idx], y.iloc[train_idx])
        calibrated[val_idx] = iso.transform(raw[val_idx])

    oofs_train[col] = calibrated

    final_iso = IsotonicRegression(out_of_bounds="clip")
    final_iso.fit(raw, y)
    oofs_test[col] = final_iso.transform(oofs_test[col])
    calibrators[col] = final_iso
```
